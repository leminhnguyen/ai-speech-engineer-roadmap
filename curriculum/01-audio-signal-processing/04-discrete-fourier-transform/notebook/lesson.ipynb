# The Discrete Fourier Transform

Run the O(N²) definition next to NumPy's FFT, then move a tone between bins to see leakage.

```python
import sys
sys.path.insert(0, '../code')
import numpy as np
from main import bin_frequencies, dft, frequency_resolution, idft, magnitude_spectrum, sine_wave

sample_rate, seconds = 8_000, 0.25
signal = sine_wave(440.0, sample_rate, seconds, 0.5) + sine_wave(1_000.0, sample_rate, seconds, 0.3)
spectrum = dft(signal)
float(np.max(np.abs(spectrum - np.fft.fft(signal)))), float(np.max(np.abs(idft(spectrum).real - signal)))
```

```python
magnitude = magnitude_spectrum(signal)
frequencies = bin_frequencies(len(signal), sample_rate)[: len(magnitude)]
top = np.argsort(magnitude)[-2:][::-1]
print('bin spacing', frequency_resolution(len(signal), sample_rate), 'Hz')
list(zip(frequencies[top], np.round(magnitude[top], 3)))
```

Change 440 Hz to 442 Hz. The bins are 4 Hz apart, so the tone now sits between two bins. What happens to the peak height, and how many bins carry energy above 0.01?

```python
import time

long_signal = sine_wave(440.0, sample_rate, 1.0)
start = time.perf_counter(); dft(long_signal); slow = time.perf_counter() - start
start = time.perf_counter(); np.fft.fft(long_signal); fast = time.perf_counter() - start
print(f'definition: {slow:.3f} s   np.fft.fft: {fast * 1e3:.3f} ms for N = {len(long_signal)}')
```

The definition builds an N × N matrix, so one second of audio already costs 64 million complex multiplications. The FFT gives the same numbers in N log N steps, which is why every audio library uses it.
