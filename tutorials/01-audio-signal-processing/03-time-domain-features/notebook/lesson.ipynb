# Time-Domain Features

Run the implementation, then change `frame_length` and `hop_length` to see how the feature sequence changes.

```python
import sys
sys.path.insert(0, '../code')
import numpy as np
from main import amplitude_envelope, frame_signal, rms_energy, sine_wave, zero_crossing_rate

sample_rate, frame_length, hop_length = 16_000, 1024, 512
signal = np.concatenate([sine_wave(200.0, sample_rate, 0.5, amplitude=0.2), sine_wave(2_000.0, sample_rate, 0.5, amplitude=0.8)])
frame_signal(signal, frame_length, hop_length).shape
```

```python
envelope = amplitude_envelope(signal, frame_length, hop_length)
rms = rms_energy(signal, frame_length, hop_length)
zcr = zero_crossing_rate(signal, frame_length, hop_length)
np.round(np.stack([envelope, rms, zcr])[:, ::4], 3)
```

Compare with `librosa.feature.rms` and `librosa.feature.zero_crossing_rate`. Try `center=False` so librosa frames the signal the same way as `frame_signal`.

```python
import librosa

lib_rms = librosa.feature.rms(y=signal, frame_length=frame_length, hop_length=hop_length, center=False)[0]
lib_zcr = librosa.feature.zero_crossing_rate(signal, frame_length=frame_length, hop_length=hop_length, center=False)[0]
float(np.max(np.abs(lib_rms - rms[:len(lib_rms)]))), float(np.max(np.abs(lib_zcr - zcr[:len(lib_zcr)])))
```
